```python
import pandas as pd
from nltk.sentiment.vader import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()
```

```python
data = pd.read_csv("reviews_dataset.csv", sep=',')

data.head()
```

```python
# Define functions for semtiment analysis

def get_sentiment(text):
    sentiment = analyzer.polarity_scores(text)
    return sentiment

# Apply VADER to the 'reviews.text' column and store the results in a new column 'sentiment'

data['sentiment'] = data['reviews.text'].apply(get_sentiment)
```

```python
# Extract compound sentiment scores

data['compound'] = data['sentiment'].apply(lambda x: x['compound'])

data['sentiment']
```

```python
# Define function for sentiment classification of reviews

def classify_sentiment(compound_score):
    if compound_score > 0.4:
        return "positive"
    elif compound_score < 0.1:
        return "negative"
    else:
        return "neutral"

data['sentiment_category'] = data['compound'].apply(classify_sentiment)
data.head()
```

```python
# Save the DataFrame to a CSV file
data.to_csv('reviews_analysed.csv', index=False)
```
