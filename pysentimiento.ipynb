```python
from pysentimiento import create_analyzer
import json
import csv
```

```python
analyzer = create_analyzer(task="sentiment", lang="en")

output_file = "sentiment_results.csv"

with open(output_file, 'w', newline='') as csv_file:
    csv_writer = csv.writer(csv_file)

    csv_writer.writerow(["Review Text", "Sentiment"])

    with open('Appliances.json', 'r') as file:
        for line in file:
            review_data = json.loads(line)
            review_text = review_data["reviewText"]

            sentiment_result = analyzer.predict(review_text)
            csv_writer.writerow([review_text, sentiment_result])
```

# Most common words (by POS)

```python
import csv
from collections import Counter
import re
import nltk
from nltk import sent_tokenize, word_tokenize, pos_tag
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

csv_file_name = "sentiment_results.csv"

nltk.download("averaged_perceptron_tagger")
nltk.download('punkt')
nltk.download('wordnet')
```

```python
positive_adjectives = []
positive_verbs = []
positive_nouns = []
positive_pronouns = []
total = 0

with open(csv_file_name, 'r', newline='') as csv_file:
    csv_reader = csv.reader(csv_file, delimiter=',')
    next(csv_reader)

    for row in csv_reader:
        review_text = row[0]
        sentiment_data = row[1]

        if 'output=POS' in sentiment_data:
            total += 1
            words = word_tokenize(review_text)
            tagged_words = pos_tag(words)
            avoid_verbs = "is, was, be, do, does, did, are, were, have, had, has, 's, been, am"

            for word, pos in tagged_words:
                word = word.lower()
                if pos.startswith("JJ"):  # Adjective
                    positive_adjectives.append(word)
                elif pos.startswith("VB"):  # Verb
                  if word not in avoid_verbs:
                      lemma = lemmatizer.lemmatize(word, pos='v')
                      positive_verbs.append(lemma)
                elif pos.startswith("NN"):  # Noun
                    positive_nouns.append(word)
                elif pos.startswith("PRP"):  # Pronoun
                    positive_pronouns.append(word)


def get_most_common_words(word_list, pos_name, top_n=10):
    most_common_words = Counter(word_list).most_common(top_n)
    print(f"\nMost Common Positive {pos_name}:")
    for word, count in most_common_words:
        print(f"{word}: {count}")

print(f"Total Positive Reviews: {total}")
get_most_common_words(positive_adjectives, "Adjectives")
get_most_common_words(positive_verbs, "Verbs")
get_most_common_words(positive_nouns, "Nouns")
get_most_common_words(positive_pronouns, "Pronouns")
```

```python
negative_adjectives = []
negative_verbs = []
negative_nouns = []
negative_pronouns = []
total = 0

with open(csv_file_name, 'r', newline='') as csv_file:
    csv_reader = csv.reader(csv_file, delimiter=',')
    next(csv_reader)

    for row in csv_reader:
        review_text = row[0]
        sentiment_data = row[1]

        if 'output=NEG' in sentiment_data:
            total+=1
            words = word_tokenize(review_text)
            tagged_words = pos_tag(words)
            avoid_verbs = "is, was, be, do, does, did, are, were, have, had, has, 's, been, am"

            for word, pos in tagged_words:
                word = word.lower()
                if pos.startswith("JJ"):  # Adjective
                    negative_adjectives.append(word)
                elif pos.startswith("VB"): # Verb
                  if word not in avoid_verbs:
                      lemma = lemmatizer.lemmatize(word, pos='v')
                      negative_verbs.append(lemma)
                elif pos.startswith("NN"):  # Noun
                    negative_nouns.append(word)
                elif pos.startswith("PRP"):  # Pronoun
                    negative_pronouns.append(word)


def get_most_common_words(word_list, pos_name, top_n=10):
    most_common_words = Counter(word_list).most_common(top_n)
    print(f"\nMost Common Negative {pos_name}:")
    for word, count in most_common_words:
        print(f"{word}: {count}")

print(f"Total Negative Reviews: {total}")
get_most_common_words(negative_adjectives, "Adjectives")
get_most_common_words(negative_verbs, "Verbs")
get_most_common_words(negative_nouns, "Nouns")
get_most_common_words(negative_pronouns, "Pronouns")
```

```python
#if you want to check for a specific word in a specific kind of sentiment review
#e.g i wanted to see why good is one of the most common words in NEG reviews: the output shows most are actually "not good"
search_word = input("Enter a word to search for in reviews: ")
sentiment = input("What kind of reviews (POS or NEG): ")

reviews_with_word = []
total = 0

with open(csv_file_name, 'r', newline='') as csv_file:
    csv_reader = csv.reader(csv_file)
    next(csv_reader)
    for row in csv_reader:
        review_text = row[0]
        sentiment_data = row[1]

        if f'output={sentiment}' in sentiment_data:
            if re.search(rf'\b{re.escape(search_word)}\b', review_text, flags=re.IGNORECASE):
              reviews_with_word.append(review_text)
              total += 1

if reviews_with_word:
    print(f"Total number of reviews containing {search_word}: {total}")
    print(f"Reviews Containing '{search_word}':")
    for review in reviews_with_word:
        print(review)
else:
    print(f"No reviews containing '{search_word}' found.")
```

```python
# I want to check the reviews where the checker was the most sure they were either POS or NEG
# hypothesis the ones it was more sure about are the most extreme

sentiment = input("What kind of reviews (POS or NEG): ")
reviews = []

with open(csv_file_name, 'r', newline='') as csv_file:
    csv_reader = csv.reader(csv_file)
    next(csv_reader)

    for row in csv_reader:
        review_text = row[0]
        sentiment_data = row[1]

        if f'output={sentiment}' in sentiment_data:
            sentiment_prob = float(sentiment_data.split(f"{sentiment}: ")[1].split(",")[0])
            if sentiment_prob > 0.9:
                reviews.append(review_text)

if reviews:
    print(f"Reviews with {sentiment} Sentiment Probability > 0.9:")
    for review in reviews:
        print(review)
else:
    print(f"No {sentiment} reviews with a sentiment probability > 0.9 found.")
```
